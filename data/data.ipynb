```python
import os
import kaggle

# The Kaggle dataset identifier
dataset_slug = 'patrutledge/uberjugaad-gmbh-enhanced-salt-dataset'

# The path to your raw data folder (assuming your notebook is inside phase1_process_mining)
target_directory = '../data/raw/'

# Ensure the target directory exists
os.makedirs(target_directory, exist_ok=True)

# Fetch and unzip the data
print(f"Downloading dataset to {target_directory}...")
kaggle.api.dataset_download_files(
    dataset_slug, 
    path=target_directory, 
    unzip=True
)
print("Download and extraction complete! Ready for processing.")
```
